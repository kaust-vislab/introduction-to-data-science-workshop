### 10. Slice the following tensor ( tensor = torch.ones(4, 4) ) and get the first row, first column the last row and last column
NumPy-like indexing and slicing is used in PyTorch
![Numpy slicing](../Images/arrayslicing.png)
(Image Source: https://lectures.scientific-python.org/intro/numpy/array_object.html#indexing-and-slicing)

````python
# CODE HERE

````

<details><summary><b>Solution</b></summary>
   <pre>

```python
tensor = torch.ones(4, 4)
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last row: {tensor[-1]}")
print(f"Last column: {tensor[..., -1]}")
tensor[:,1] = 0
print(tensor)
```
