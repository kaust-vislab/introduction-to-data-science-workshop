# Data Handling with PyTorch

## Why Do We Need a DataLoader?

When working with large datasets, it is essential to efficiently load and preprocess the data. PyTorch provides the `DataLoader` class to handle this task. The `DataLoader` object serves several important purposes:

1. **Batching**: It allows you to divide your dataset into smaller batches, which helps in efficient training and reduces memory usage.
2. **Shuffling**: It can shuffle the data at every epoch, which helps in breaking the correlation between consecutive samples and improves the generalization of the model.
3. **Parallelism**: It can load data in parallel using multiple worker processes, which speeds up the data loading process.
4. **Data Augmentation**: It can apply transformations and augmentations to the data on-the-fly, which helps in increasing the diversity of the training data.

By using a `DataLoader`, you can streamline the process of loading, preprocessing, and augmenting your data, making it easier to train your models efficiently.

## Loading Datasets

PyTorch provides a powerful library called `torchvision` that contains popular datasets, model architectures, and image transformations. Let's start by loading the FashionMNIST dataset.

```python
import torch
from torchvision import datasets, transforms

# Define a transform to normalize the data
transform = transforms.Compose([transforms.ToTensor()])

# Download and load the training data
trainset = datasets.FashionMNIST(root='./data', 
                                train=True, 
                                download=True, 
                                transform=transform
                                )
trainloader = torch.utils.data.DataLoader(trainset, 
                                        batch_size=64, 
                                        shuffle=True
                                        )

# Download and load the test data
testset = datasets.FashionMNIST(root='./data', 
                                train=False, 
                                download=True, 
                                transform=transform
                                )
testloader = torch.utils.data.DataLoader(testset, 
                                         batch_size=64, 
                                         shuffle=False
                                         )
```
